```python
from langchain_openrouter import ChatOpenRouter
```

```python
from dotenv import load_dotenv
import os

load_dotenv()
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")

if not openrouter_api_key:
    raise RuntimeError("Open router api key missing")

model = "nvidia/nemotron-3.5-lightning:free"
llm = ChatOpenRouter(
    model=model,
    temperature=0,
    api_key=openrouter_api_key,
)
```

```python
SYSTEM = """You are the assistant for Sweet Crumbs Home Bakery.
Facts: we bake to order; orders need 24 hours' notice; all cakes have eggless versions.
Rules: never promise delivery times; if you do not know, say so.
Format: reply politely in under 60 words."""
```

```python
def ask_question(question):
    response = llm.invoke([("system", SYSTEM), ("human", question)]).content
    return response
```

```python
print(ask_question("Can I get a chocolate cake for tomorrow?"))
```

```python
from openai import OpenAI

client = OpenAI(
    api_key=openrouter_api_key,
    base_url="https://openrouter.ai/api/v1",
)
completion = client.chat.completions.create(
    model=model,
    messages=[
        {
            "role": "user",
            "content": "Explain how to keep API keys secure."
        }
    ]
)
print(completion.choices[0].message.content)
```
